```python
import pandas as pd
from IPython.display import display, HTML

# 1. Load the Excel data file
excel_file = "chigumira data.xlsx"
data = pd.read_excel(excel_file)

# 2. Display the first 10 rows as a beautiful HTML preview table
display(HTML(data.head(10).to_html(index=False, classes='table table-striped')))
```

```python
# Generate an SPSS-style descriptives summary table
spss_descriptive_table = data.describe().T[['count', 'mean', 'std', 'min', 'max']]
spss_descriptive_table.columns = ['N', 'Mean', 'Std. Deviation', 'Minimum', 'Maximum']

# Display the formatted table
display(HTML(spss_descriptive_table.to_html(classes='table table-bordered')))
```

```python
import pandas as pd
import numpy as np
from IPython.display import display, HTML

# 1. Select your target variable (Change "a1" to "a2", "a3", etc. to test others)
target_column = "a1" 

# 2. Get the unique values sorted numerically (ignoring NaN for sorting)
unique_vals = sorted([x for x in data[target_column].unique() if pd.notna(x)])
if data[target_column].isna().any():
    unique_vals.append(np.nan)

# 3. Calculate components exactly like SPSS
total_cases = len(data)
missing_count = data[target_column].isna().sum()
valid_total = total_cases - missing_count

rows = []
cumulative_pct = 0.0

for val in unique_vals:
    if pd.isna(val):
        continue
    freq = int((data[target_column] == val).sum())
    pct = (freq / total_cases) * 100
    valid_pct = (freq / valid_total) * 100
    cumulative_pct += valid_pct
    
    rows.append({
        'Label': int(val), 'Type': 'Valid', 'Frequency': freq, 
        'Percent': pct, 'Valid Percent': valid_pct, 'Cumulative Percent': cumulative_pct
    })

# Add Valid Total Row
rows.append({
    'Label': 'Total', 'Type': 'Valid_Total', 'Frequency': valid_total, 
    'Percent': (valid_total / total_cases) * 100, 'Valid Percent': 100.0, 'Cumulative Percent': np.nan
})

# Add System Missing Row if it exists
if missing_count > 0:
    rows.append({
        'Label': 'System', 'Type': 'Missing', 'Frequency': missing_count, 
        'Percent': (missing_count / total_cases) * 100, 'Valid Percent': np.nan, 'Cumulative Percent': np.nan
    })

# Add Grand Total Row
rows.append({
    'Label': 'Total', 'Type': 'Grand_Total', 'Frequency': total_cases, 
    'Percent': 100.0, 'Valid Percent': np.nan, 'Cumulative Percent': np.nan
})

spss_df = pd.DataFrame(rows)

# 4. Generate the exact SPSS Visual Layout using HTML/CSS
spss_html = f"""
<div style="font-family: 'Times New Roman', Times, serif; color: black; margin-bottom: 25px;">
    <table style="border-collapse: collapse; width: 550px; font-size: 13px; text-align: right; border-top: 2px solid black; border-bottom: 2px solid black;">
        <caption style="text-align: left; font-weight: bold; font-size: 14px; padding-bottom: 8px; color: black;">{target_column}</caption>
        <thead>
            <tr style="border-bottom: 1px solid black;">
                <th style="text-align: left; padding: 4px; min-width: 100px;"></th>
                <th style="text-align: left; padding: 4px; min-width: 80px;"></th>
                <th style="padding: 4px; font-weight: normal;">Frequency</th>
                <th style="padding: 4px; font-weight: normal;">Percent</th>
                <th style="padding: 4px; font-weight: normal;">Valid Percent</th>
                <th style="padding: 4px; font-weight: normal;">Cumulative Percent</th>
            </tr>
        </thead>
        <tbody>
"""

# Track if we have already printed the side-labels for "Valid" or "Missing" grouping blocks
valid_labeled = False
missing_labeled = False

for idx, row in spss_df.iterrows():
    # Formatting row borders and font weights to look like a real SPSS canvas printout
    style = "padding: 3px 5px;"
    font_weight = "normal"
    border_bottom = "none"
    
    if row['Type'] in ['Valid_Total', 'Grand_Total']:
        font_weight = "bold"
        style += " border-top: 1px solid black;"
    if row['Type'] in ['Valid_Total', 'Missing']:
        border_bottom = "1px solid black"
        
    spss_html += f"<tr style='font-weight: {font_weight}; border-bottom: {border_bottom};'>"
    
    # Leftmost structural grouping columns
    if row['Type'] in ['Valid', 'Valid_Total']:
        if not valid_labeled:
            spss_html += f"<td style='text-align: left; padding: 3px 5px;' rowspan='{len(spss_df[spss_df['Type'].isin(['Valid', 'Valid_Total'])])}'>Valid</td>"
            valid_labeled = True
    elif row['Type'] == 'Missing':
        spss_html += f"<td style='text-align: left; padding: 3px 5px;' rowspan='{len(spss_df[spss_df['Type'] == 'Missing'])}'>Missing</td>"
    elif row['Type'] == 'Grand_Total':
        spss_html += "<td></td>"

    # Print the specific Row values
    lbl = row['Label']
    f_val = f"{int(row['Frequency'])}"
    p_val = f"{row['Percent']:.1f}"
    v_val = "" if pd.isna(row['Valid Percent']) else f"{row['Valid Percent']:.1f}"
    c_val = "" if pd.isna(row['Cumulative Percent']) else f"{row['Cumulative Percent']:.1f}"
    
    spss_html += f"""
        <td style="text-align: left; {style}">{lbl}</td>
        <td style="{style}">{f_val}</td>
        <td style="{style}">{p_val}</td>
        <td style="{style}">{v_val}</td>
        <td style="{style}">{c_val}</td>
    </tr>
    """

spss_html += "</tbody></table></div>"
display(HTML(spss_html))
```
